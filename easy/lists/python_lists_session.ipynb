````python
# Q1: What is the output of this slice?
# Write your prediction, then run to check. [2, 4]

print([1, 2, 3, 4, 5][1:-1:2])
````

<details><summary>Answer</summary>

````python
# Q2: After these operations, what is a?
# Prediction: [1, 2, 3, 4] because a is equal to b therefore the same

a = [1, 2, 3]
b = a #points to the same address in memory
b += [4]
print(a)
````

<details><summary>Answer</summary>

`[1, 2, 3, 4]` — `b = a` makes b an alias. `+=` on a list calls `extend` in-place, mutating the shared object. Both a and b become `[1, 2, 3, 4]`.
</details>

````python
# Q3: What does x contain?
# Prediction: None

x = [3, 1, 2].sort()
print(x)
````

<details><summary>Answer</summary>

`None` — `.sort()` modifies in-place and returns `None`. The temporary list is created, sorted, then discarded. `x` stores the return value of `.sort()`, which is `None`.
</details>

````python
lst = [1, 2, 3]
new_list = lst[:] # lst: [1, 2, 3] new_lst: [1, 2, 3, 99]

new_list.append(29)
print(lst, new_list)
````

````python
# Q4: Which of these creates a real independent copy?
# Run each option and verify by modifying new_list and checking if lst changes.

lst = [1, 2, 3]

# Option A - no
new_list = lst  # lst: [1, 2, 3, 99] new_lst: [1, 2, 3, 99]

# Option B - yes
new_list = lst[:] # lst: [1, 2, 3] new_lst: [1, 2, 3, 99]
lst = [1, 2, 3]
# Option C - yes
new_list = lst.copy() # lst: [1, 2, 3] new_lst: [1, 2, 3, 99]

# Option D - yes
new_list = list(lst) # lst: [1, 2, 3] new_lst: [1, 2, 3, 99]

new_list.append(99)
print("lst:", lst)          # did lst change? No
print("new_list:", new_list)
````

<details><summary>Answer</summary>

B, C, and D all create independent copies. Only **A** (`new_list = lst`) creates an alias — mutating `new_list` will mutate `lst` too. `lst[:]`, `.copy()`, and `list(lst)` are all equivalent shallow copies.
</details>

````python
# Q5: What is len(a) after this?
# Prediction: 3 because ([3, 4]) is a list in list a

a = [1, 2]
a.append([3, 4])
print(len(a))
print(a)
````

`3` — `.append()` adds `[3, 4]` as a **single item**, resulting in `[1, 2, [3, 4]]`. That's 3 items. Use `.extend()` to add them flat.
</details>
