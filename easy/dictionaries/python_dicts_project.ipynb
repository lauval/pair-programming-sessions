Let's build the core functions. Each one has hints to guide you.

````python
def add_movie(collection, title, year, genre):
    """
    Add a new movie to the collection.
    New movies always start with status "watchlist" and rating None.
    If the movie already exists, print a message and don't overwrite it.
    """
    # Hint: check whether title is already a key in collection
    if title not in movies.keys():
        print(f"'{title}' is not in your collection.")
        collection[title] = {"year" : year , "genre" : genre, "status" : "watchlist", "rating" : None}
        print(f"Added '{title}' to your watchlist.")
        
    else:
        print(f"'{title}' is already in your collection.")
    
    

    # Hint: assign a new dict to collection[title]
    #collection[title] = {}
    #print(f"Added '{title}' to your watchlist.")


# Test it
add_movie(movies, "Amélie", 2001, "Romance")
add_movie(movies, "Inception", 2010, "Sci-Fi")   # should print a warning
print(movies["Amélie"])
````

````python
def mark_watched(collection, title, rating):
    """
    Mark a movie as watched and record a rating (1–10).
    If the movie isn't in the collection, print a message.
    If the rating is outside 1–10, print a message and don't update.
    """
    # Hint: use .get() to safely check if the movie exists
    movie = collection.get(title)
    if movie is None:
        print(f"'{title}' not found. Add it first.")
        return

    if rating not in range(0,11):   # Hint: check the rating range
        print("Rating must be between 1 and 10.")
        return

    # Hint: update the two relevant fields inside the movie dict
    movie["status"] = "watched"
    movie["rating"] = rating
    print(f"Marked '{title}' as watched ({rating}/10).")


# Test it
mark_watched(movies, "Parasite", 10)
mark_watched(movies, "Amélie", 9)
mark_watched(movies, "Amélie", 11)    # should print a warning
mark_watched(movies, "Interstellar", 8)  # should say not found
print(movies["Parasite"])
````

````python
def lookup(collection, title):
    """
    Print a summary of a single movie.
    If not found, print a helpful message.

    Output should look like:
        Inception (2010) — Sci-Fi
        Status: watched | Rating: 8/10

    For an unwatched film, show "not rated yet" instead of the rating.
    """
    # Hint: use .get() and handle the None case
    movie = movies.get(title)

    if movie is None:
        print(f"Movie {title} is not in the collection, please add it first")
    else:
        rating_str = movie["rating"] if movie["rating"] else "not rated yet"
        print(f"{title} ({movie['year']}) - ({movie['genre']})")
        print(f"Status: {movie['status']} | Rating: {rating_str}")


# Test it
lookup(movies, "Inception")
lookup(movies, "Dune: Part Two")
lookup(movies, "The Godfather")   # not in collection
````

````python
def get_watchlist(collection):
    """
    Return a list of titles that have status "watchlist".
    Print each title, or a message if the watchlist is empty.
    """
    watchlist = []
    for title, value in movies.items():
        if value["status"] == "watchlist":
            watchlist.append(title)
            
    if watchlist:
        for movie in watchlist:
            print(movie)
    else:
        print("Watchlist is empty.")
    return


# Test it
get_watchlist(movies)
````

````python
def by_genre(collection, genre):
    """
    Print all movies in the given genre.
    Include their status and rating in the output.
    Print a message if none are found.
    The genre match should be case-insensitive.
    """
    genre_list = []
    for movie in collection.keys():
        genre_list.append(collection[movie]["genre"])
    if  genre in genre_list:
        for movie in collection.keys():
            if collection[movie]["genre"] == genre:
                movie_name = movie
                status = collection[movie]["status"]
                ratings = collection[movie]["rating"]
                print(f" Movie title: {movie_name} \n status: {status}, ratings: {ratings}")
    else:
        print("Genre does not exist, no movie found")
# Test it
by_genre(movies, "Sci-Fi")
by_genre(movies, "Horror")
````

````python
def top_rated(collection, n=3):
    """
    Print the top n watched movies by rating, highest first.
    Only include movies that have been watched (rating is not None).
    """    
    rated_movies = []
    
    for title, value in collection.items():
        if value["rating"] is not None:
            rated_movies.append((title, value["rating"]))

    rated_movies.sort(key=lambda movie: movie[1], reverse=True)

    for title, rating in rated_movies[:n]:
        print(f"{title} - {rating}/10")

    print()
# Test it
top_rated(movies)
top_rated(movies, n=2)
````

````python
# Write a function called average_rating(collection).
# It should compute and print the average rating across all watched movies.
# Round to 1 decimal place.
# If no movies have been watched, print a suitable message.
def average_rating(collection):
    avg_ratings = []
    for title, value in movies.items():
        if value["rating"] is not None:
            avg_ratings.append(value["rating"])
    if len(avg_ratings) == 0:
        print("No watched movies yet.")
    else:
        average = sum(avg_ratings) / len(avg_ratings)

        print(f" Average Ratings : {round(average, 1)}")
average_rating(movies)
````

````python
# Write a function called genre_breakdown(collection).
# It should print how many movies exist per genre, sorted by count (highest first).
# Example output:
#   Sci-Fi  : 3
#   Drama   : 2
#   Comedy  : 1
def genre_breakdown(collection):
    genres = {}
    for title, value in collection.items():

        genre = value["genre"]

        if genre not in genres:
            genres[genre] = 1
        else:
            genres[genre] += 1

    sorted_genres = sorted(
        genres.items(),
        key=lambda item: item[1],
        reverse=True
    )
    
    for genre, count in sorted_genres:

        print(f"{genre:<10} : {count}")       
genre_breakdown(movies)
````

````python
# Write a function called summary(collection).
# It should print an overview of the whole collection, something like:
#
#   📽  Movie Tracker — 7 titles
#   ✅  Watched  : 4
#   🕐  Watchlist: 3
#   ⭐  Average rating: 8.7 / 10
#   🎭  Genres: Drama, Sci-Fi, Comedy, Thriller, Romance
#
# Reuse your average_rating logic here rather than repeating it.
def average_rating(collection):

    ratings = []

    for value in collection.values():

        if value["rating"] is not None:
            ratings.append(value["rating"])

    if ratings:
        return round(sum(ratings) / len(ratings), 1)

    return None


def summary(collection):

    total_movies = len(collection)

    watched = 0
    watchlist = 0
    genres = set()

    for value in collection.values():
        if value["status"] == "watched":
            watched += 1

        elif value["status"] == "watchlist":
            watchlist += 1

        genres.add(value["genre"])

    avg = average_rating(collection)

    print(f"📽  Movie Tracker — {total_movies} titles")
    print(f"✅  Watched  : {watched}")
    print(f"🕐  Watchlist: {watchlist}")

    if avg is not None:
        print(f"⭐  Average rating: {avg} / 10")
    else:
        print("⭐  Average rating: No ratings yet")

    print(f"🎭  Genres: {', '.join(sorted(genres))}")
summary(movies)
````

````python
def recommend(collection):

    highest_rating = 0
    favorite_genres = set()

    # Find highest rating
    for value in collection.values():

        if value["rating"] is not None:

            if value["rating"] > highest_rating:
                highest_rating = value["rating"]

    # Find genres of highest-rated movies
    for value in collection.values():

        if value["rating"] == highest_rating:
            favorite_genres.add(value["genre"])

    recommendations = []

    # Find watchlist movies in favorite genres
    for title, value in collection.items():

        if (
            value["status"] == "watchlist"
            and value["genre"] in favorite_genres
        ):

            recommendations.append(title)

    # Print recommendations
    if recommendations:

        print("Recommended from your favorite genres:\n")

        for movie in recommendations:
            print(movie)

    else:
        print("No watchlist movies found in your favorite genres.")
````
