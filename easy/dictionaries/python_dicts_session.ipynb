|---|---|
| Structure | Key → Value pairs |
| Ordered? | ✅ Yes (insertion order, Python 3.7+) |
| Allows duplicate keys? | ❌ No — later value wins |

## Creating Dictionaries — Warm-up
Run these cells and observe before moving to the activities.

````python
# Literal syntax
person = {"name": "Alice", "age": 30, "city": "London"}
print(person)

# dict() constructor with keyword arguments
config = dict(host="localhost", port=8080, debug=True)
print(config)

# dict() from a list of (key, value) tuples
pairs = dict([("a", 1), ("b", 2), ("c", 3)])
print(pairs)

# fromkeys() — same value for multiple keys
scores = dict.fromkeys(["Alice", "Bob", "Carol"], 0)
print(scores)   # {'Alice': 0, 'Bob': 0, 'Carol': 0}
````

````python
# Duplicate keys — last value wins, no error raised
d = {"x": 1, "x": 2, "x": 3}
print(d)   # {'x': 3}

# Accessing values
print(person["name"])         # 'Alice'  — raises KeyError if missing
print(person.get("age"))      # 30
print(person.get("email"))    # None     — safe, no error
print(person.get("email", "N/A"))  # 'N/A'  — with a default
````

## Core Methods — Reference

| Method | What it does | Returns |
|---|---|---|
| `d[key] = val` | Add or update a key | — |
| `d.get(key, default)` | Safe value lookup | value or default/None |
| `d.update(other)` | Merge another dict in (in-place) | None |
| `d.pop(key, default)` | Remove key, return its value | value |
| `d.popitem()` | Remove & return last inserted (key, value) | tuple |
| `d.setdefault(key, val)` | Return value if key exists, else insert & return default | value |
| `d.keys()` | View of all keys | dict_keys |
| `d.values()` | View of all values | dict_values |
| `d.items()` | View of all (key, value) pairs | dict_items |
| `d.copy()` | Shallow copy | dict |
| `d.clear()` | Remove all entries | None |

````python
# Iterating — the three views
inventory = {"apples": 5, "bananas": 3, "cherries": 12}

for key in inventory.keys():
    print(key)

for value in inventory.values():
    print(value)

for key, value in inventory.items():   # most common pattern
    print(f"{key}: {value}")
````

## ⭐ Advanced — Dictionary Comprehensions

Just like list comprehensions, dict comprehensions build a dictionary from an iterable in a single expression.

**Syntax:**
```python
{key_expr: value_expr for item in iterable if condition}
```

The `if condition` is optional — it filters which items get included.

````python
# Basic: square each number
squares = {x: x**2 for x in range(1, 6)}
print(squares)   # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# With a filter: only even squares
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)   # {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}

# Transform an existing dict — apply a discount to all prices
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00}
discounted = {item: round(price * 0.9, 2) for item, price in prices.items()}
print(discounted)

# Invert a dict — swap keys and values
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(inverted)   # {1: 'a', 2: 'b', 3: 'c'}

# From two lists using zip()
keys   = ["name", "age", "city"]
values = ["Alice", 30, "London"]
record = {k: v for k, v in zip(keys, values)}
print(record)
````

````python
# Q7: You want a copy of d that won't affect the original when modified.
# Which line gives you a true independent copy?

d = {"name": "Alice", "score": 100}

copy_a = d               # option A
copy_b = d.copy()        # option B
copy_c = dict(d)         # option C

# Modify each copy and check if d changed
copy_a["score"] = 0
print("After copy_a mutation:", d)

d = {"name": "Alice", "score": 100}  # reset
copy_b["score"] = 0
print("After copy_b mutation:", d)

d = {"name": "Alice", "score": 100}  # reset
copy_c["score"] = 0
print("After copy_c mutation:", d)
````

<details><summary>Answer</summary>

Only **A** mutates the original. Both `.copy()` and `dict(d)` produce independent shallow copies. This mirrors the list aliasing gotcha — assignment always copies the reference, never the object.
</details>
