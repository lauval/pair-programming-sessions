## Creating Strings — Warm-up

````python
# Single and double quotes are interchangeable
a = 'hello'
b = "hello"
print(a == b)   # True

# Use one to avoid escaping the other
quote  = "It's a beautiful day"
dialog = 'She said "hello" and walked away'

# Multiline strings
poem = """
Roses are red,
Violets are blue,
Python is fun,
And so are you.
"""
print(poem)

# Raw strings — backslashes are treated as literals
# Useful for file paths and (later) regular expressions
path  = r"C:\Users\name\Documents"
print(path)
````

````python
# Indexing and slicing work exactly like lists
s = "Python"
#    012345
#   -654321

print(s[0])     # 'P'
print(s[-1])    # 'n'
print(s[1:4])   # 'yth'
print(s[::-1])  # 'nohtyP'

# But strings are IMMUTABLE — you cannot assign to an index
try:
    s[0] = "J"
except TypeError as e:
    print(f"Error: {e}")

# To "change" a string, you create a new one
s = "J" + s[1:]
print(s)   # 'Jython'
````

## Core Methods — Reference

All string methods return a **new string** — they never modify the original.

### Case
| Method | Example | Result |
|---|---|---|
| `.upper()` | `"hello".upper()` | `"HELLO"` |
| `.lower()` | `"HELLO".lower()` | `"hello"` |
| `.title()` | `"hello world".title()` | `"Hello World"` |
| `.capitalize()` | `"hello world".capitalize()` | `"Hello world"` |
| `.swapcase()` | `"Hello".swapcase()` | `"hELLO"` |

### Search
| Method | What it does | Returns |
|---|---|---|
| `.find(sub)` | Index of first occurrence, -1 if missing | int |
| `.count(sub)` | Number of non-overlapping occurrences | int |
| `.startswith(prefix)` | Starts with prefix? | bool |
| `.endswith(suffix)` | Ends with suffix? | bool |
| `sub in s` | Membership test | bool |

### Transform
| Method | What it does |
|---|---|
| `.replace(old, new)` | Replace all occurrences of old with new |
| `.strip()` | Remove leading and trailing whitespace (or given chars) |
| `.lstrip()` / `.rstrip()` | Strip left or right side only |
| `.split(sep)` | Split into a list on sep (default: any whitespace) |
| `sep.join(iterable)` | Join an iterable into a string with sep between each item |
| `.zfill(width)` | Pad with leading zeros to reach width |

### Inspection
| Method | True when... |
|---|---|
| `.isalpha()` | All characters are letters |
| `.isdigit()` | All characters are digits |
| `.isalnum()` | All letters or digits, no spaces |
| `.isspace()` | Only whitespace |
| `.isupper()` / `.islower()` | All cased characters are upper/lower |

## String Formatting

There are three ways to format strings in Python. **f-strings are the modern standard.**

````python
name  = "Alice"
score = 88.5
rank  = 3

# %-formatting (old, avoid in new code)
print("Name: %s, Score: %.1f" % (name, score))

# .format() (Python 3, still common)
print("Name: {}, Score: {:.1f}".format(name, score))

# f-strings (Python 3.6+, preferred)
print(f"Name: {name}, Score: {score:.1f}")

# f-strings can contain any expression
print(f"{'Passed' if score >= 50 else 'Failed'} — rank #{rank}")
print(f"{name.upper():>20}")   # right-aligned in 20-char field
print(f"{score:08.2f}")        # 0-padded, 8 wide, 2 decimal places
print(f"{1_000_000:,}")        # thousands separator
````

## ⭐ Regular Expressions — An Introduction

A **regular expression** (regex) is a pattern that describes a set of strings. Instead of checking if a string equals a fixed value, you check whether it *matches a shape*.

Python's `re` module is the standard library for this.

```python
import re
```

### Reading the syntax

| Pattern | Matches |
|---|---|
| `abc` | The literal characters `abc` |
| `.` | Any single character (except newline) |
| `\d` | Any digit — same as `[0-9]` |
| `\w` | Any word character — letters, digits, underscore |
| `\s` | Any whitespace character — space, tab, newline |
| `\D`, `\W`, `\S` | The OPPOSITE of the above |
| `[abc]` | Any one of: a, b, or c |
| `[a-z]` | Any lowercase letter |
| `[^abc]` | Anything EXCEPT a, b, or c |
| `^` | Start of the string |
| `$` | End of the string |
| `*` | 0 or more of the preceding |
| `+` | 1 or more of the preceding |
| `?` | 0 or 1 of the preceding (optional) |
| `{n}` | Exactly n of the preceding |
| `{n,m}` | Between n and m of the preceding |
| `(abc)` | A group — treat `abc` as a unit |
| `a\|b` | Either `a` or `b` |

### The three functions you'll use most

| Function | What it does |
|---|---|
| `re.search(pattern, string)` | Find the **first** match anywhere in the string. Returns a match object or `None`. |
| `re.findall(pattern, string)` | Return a **list of all** matching substrings. |
| `re.sub(pattern, replacement, string)` | Replace every match with the replacement string. |

> **Always use raw strings for patterns** — `r"\d+"` not `"\d+"` — to avoid Python interpreting the backslashes before the regex engine sees them.

### Reading Regex — What does each pattern match?

````python
import re

# Before running each line, write what you think it will find.

samples = [
    "My number is 07700 900123",
    "Contact us at hello@example.com",
    "Order #00142 placed on 2024-03-15",
    "Price: £4.99 (was £9.99)",
]

for s in samples:
    # Pattern 1: one or more digits
    print(re.findall(r"\d+", s))

print()
for s in samples:
    # Pattern 2: a word character sequence of exactly 3 letters
    print(re.findall(r"\b\w{3}\b", s))

print()
for s in samples:
    # Pattern 3: anything that looks like a date (YYYY-MM-DD)
    print(re.findall(r"\d{4}-\d{2}-\d{2}", s))
````

Breaking it down:

### Applying Regex to a Collection

Earlier in the dictionaries session you built a word frequency counter. Here's a similar problem — but this time you'll use regex to filter and transform a messy list of strings before processing them.

````python
import re

# A messy list of user-submitted tags (some have junk characters, some are numbers,
# some are fine). We want only tags that are purely alphabetic words (no digits,
# no symbols, no spaces) and at least 3 characters long.

raw_tags = [
    "python", "data123", "ML", "ai", "!!hot", "web",
    "deep-learning", "NLP", "42", "cloud", "dev ops", "rust",
    "c++", "go", "kotlin", "#trending", "sql", "ab", "typescript"
]

# Step 1: Write a pattern that matches only purely alphabetic tags
#         that are at least 3 characters long.
pattern = r"___"

# Step 2: Filter the list
clean_tags = [tag for tag in raw_tags if re.fullmatch(pattern, tag)]
print("Clean tags:", clean_tags)

# Step 3: Build a frequency dict from the clean tags (normalised to lowercase)
freq = {}
for tag in clean_tags:
    tag = tag.lower()
    freq[tag] = freq.get(tag, 0) + 1

print("\nTag frequencies:")
for tag, count in sorted(freq.items(), key=lambda x: x[1], reverse=True):
    print(f"  {tag}: {count}")
````


`re.fullmatch()` requires the pattern to match the **entire** string, not just part of it — so `data123` fails because `123` isn't covered, and `de-learning` fails because `-` isn't in `[A-Za-z]`.

Expected clean tags: `['python', 'ML', 'web', 'NLP', 'cloud', 'rust', 'kotlin', 'sql', 'typescript']`

> Note: `"go"` and `"ab"` are filtered out (too short). `"dev ops"` fails because of the space. `"deep-learning"` fails because of the hyphen.
