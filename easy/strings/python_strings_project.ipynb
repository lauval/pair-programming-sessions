## Part 1 — Cleaning

Before any analysis, we need the posts in a consistent shape. Each function has its logic sketched out — fill in the blanks.

````python
def clean_post(post):
    """
    Return a cleaned version of a single post:
    - Strip leading and trailing whitespace
    - Collapse any run of internal whitespace to a single space
    - Remove emoji (for now, just strip any non-ASCII characters)
    """
    # Hint: strip() handles the edges.
    # Hint: split() then join() collapses internal whitespace.
    # Hint: to remove non-ASCII, encode to ASCII with errors='ignore', then decode.

    post = post.strip()
    post = " ".join(post.split()) # did  post = post.join(post.split(" "))
    post = post.encode("ascii", errors="ignore").decode("ascii")
    return post


def clean_all(posts):
    """Return a new list with clean_post applied to every post."""
    return [clean_post(post) for post in posts]


# Test
cleaned = clean_all(posts)
for p in cleaned:
    print(repr(p))
````

````python
def get_words(post):
    """
    Return a list of plain words from a post:
    - Lowercase everything
    - Remove punctuation attached to words (.,!?:;'")
    - Exclude tokens that start with @ or #
    - Exclude words shorter than 2 characters
    """
    punctuation = ".,!?:;'\"\u2019"

    words = []
    for token in post.lower().split():
        # Hint: skip mentions and hashtags
        if token.startswith("#") or token.startswith("@"):
            continue
        # Hint: strip punctuation from both ends of the token
        token = token.strip(punctuation)
        # Hint: only keep tokens that are at least 2 chars long
        if len(token) >= 2:
            words.append(token)
    return words


# Test
print(get_words(cleaned[0]))
````

## Part 2 — Extraction

Function signatures and docstrings are here. Implementation is up to you.

````python
def extract_mentions(post):
    """
    Return a list of all @mentions in a post, lowercased, without the @ symbol.
    E.g. '@Alice and @Bob' → ['alice', 'bob']
    Strip any trailing punctuation from the username.
    """
    punctuation = ".,!?:;'\"\u2019"
    mentions = []
    for token in post.split():
        if token.startswith("@"):
            token = token[1:].strip(punctuation)
            mentions.append(token)
    return mentions


# Test
print(extract_mentions(cleaned[0]))
print(extract_mentions(cleaned[2]))
````

````python
def extract_hashtags(post):
    """
    Return a list of all #hashtags in a post, lowercased, without the # symbol.
    E.g. '#Python #webdev' → ['python', 'webdev']
    Strip any trailing punctuation.
    """
    punctuation = ".,!?:;'\"\u2019"
    hashtags = []
    for token in post.split():
        if token.startswith("#"):
            token = token[1:].strip(punctuation)
            hashtags.append(token)
    return hashtags
    
# Test
print(extract_hashtags(cleaned[0]))
print(extract_hashtags(cleaned[4]))
````

````python
def word_frequency(posts):
    """
    Build and return a frequency dict of all plain words across every post.
    Use get_words() to tokenise each post.
    Exclude common stop words (provided below).
    """
    stop_words = {
        "the", "and", "is", "it", "in", "on", "at", "to",
        "a", "an", "of", "for", "my", "me", "so", "but",
        "you", "your", "are", "was", "how", "now", "this"
    }

    freq = {}

    for post in posts:
        for word in get_words(post):
            if word not in stop_words:
                # checks if the word has been added to the dictionary, if the word is in dictionary is skips over it
                freq[word] = freq.get(word, 0) + 1
    return freq

# Test — print the 10 most common words
freq = word_frequency(cleaned)
top10 = sorted(freq.items(), key=lambda x: x[1], reverse=True)[:10]
for word, count in top10:
    print(f"  {word:<20} {count}")
````

## Part 3 — Aggregation

No signatures. Write the whole function from the description.

````python
# Write a function top_hashtags(posts, n=5).
# It should return the n most used hashtags across all posts, as a list of
# (hashtag, count) tuples sorted by count descending.
# Use extract_hashtags() on each post.

def top_hashtags(posts, n=5):
    count_hashtags = {}

    for post in posts:
        for hashtag in extract_hashtags(post):
            count_hashtags[hashtag] = count_hashtags.get(hashtag, 0) + 1

    return sorted(
        count_hashtags.items(),
        key=lambda x: x[1],
        reverse=True
    )[:n]

print(top_hashtags(cleaned))
````

````python
# Write a function most_mentioned(posts, n=5).
# Same idea — find the n most mentioned users across all posts.
# Use extract_mentions() on each post.
def most_mentioned(post, n=5):
    count_mentions =  {}

    for post in posts:
        for mention in extract_mentions(post):
            count_mentions[mention] = count_mentions.get(mention, 0) + 1

    return sorted(
        count_mentions.items(),
        key=lambda x: x[1],
        reverse=True
    )[:n]

print(most_mentioned(cleaned))
````

````python
# Write a function posts_containing(posts, query).
# Return a list of posts (from the cleaned list) that contain the query string.
# The search should be case-insensitive.

def posts_containing(posts, query):
    matches = []

    for post in posts:
        cleaned_post = clean_post(post)
        if query.lower() in cleaned_post.lower():
            matches.append(cleaned_post)

    return matches
print(posts_containing(posts, "python"))
````

## Part 4 — The Report

Bring it all together. No hints.

````python
# Write a function generate_report(posts) that prints a formatted summary.
#
# It should look something like this (exact formatting is up to you):
#
#   ============================================
#          POST ANALYSER — 8 posts
#   ============================================
#
#   TOP HASHTAGS
#   #python           3
#   #webdev           2
#   #machinelearning  2
#   ...
#
#   MOST MENTIONED
#   @alice            2
#   @bob              2
#   ...
#
#   TOP WORDS
#   learning          3
#   ...
#
#   ============================================
#
# Use your functions from Parts 2 and 3. Use f-strings for alignment.

def generate_report(posts):
    width = 34

    print("=" * 34)
    print("POST ANALYSER - 8 posts".center(width))
    print("=" * 34)
    
    print("\nTOP HASHTAGS")
    for hashtag, count in top_hashtags(cleaned):
        print(f"#{hashtag:<16} {count}")
        
    
    print("\nMOST MENTIONED")
    for mention, count in most_mentioned(cleaned):
        print(f"@{mention:<16} {count}")

    print("\nTOP WORDS")
    for word, count in sorted(freq.items(), 
                              key=lambda x: x[1],
                              reverse=True):
        print(f"{word:<16}{count}")
    print()
    
    print("=" * 34)

generate_report(cleaned)
````

## ⭐ Bonus — Regex

Your `extract_mentions()` and `extract_hashtags()` functions in Part 2 most likely used `.split()` and `.startswith()`. That works well — but regex can do it in fewer lines and handles edge cases more robustly.

**Challenge:** Rewrite both functions using `re.findall()`. Then add a third function that extracts URLs.

````python
import re

# Rewrite extract_mentions using re.findall().
# Hint: a mention is @ followed by one or more word characters.

def extract_mentions_re(post):
    #for post in posts:
    mentions = re.findall(r"@([\w']+)", post)
    return mentions

# Rewrite extract_hashtags using re.findall().
# Hint: same idea — # followed by one or more word characters.

def extract_hashtags_re(post):
    hashtags = re.findall(r"#(\w+)", post) 
    return hashtags

#print(extract_mentions_re(post))
#print(extract_mentions(post))

# Verify your regex versions match the original ones on all cleaned posts
for post in cleaned:
    assert extract_mentions_re(post) == extract_mentions(post), f"Mismatch: {post}"
    assert extract_hashtags_re(post) == extract_hashtags(post), f"Mismatch: {post}"

print("✅ All regex versions match the originals.")
````

````python
# Now add some posts that contain URLs and write extract_urls(post).
# A URL for our purposes starts with http:// or https://, followed by non-whitespace characters.

posts_with_urls = cleaned + [
    "Great write-up on list comprehensions: https://realpython.com/list-comprehension-python/ — highly recommend",
    "Two resources that helped me this week: https://docs.python.org/3/ and http://regexr.com",
    "No links here, just vibes #python",
]


def extract_urls(post):
    """
    Return a list of all URLs in a post.
    A URL starts with http:// or https:// and continues until whitespace.
    """
    urls_found = re.findall(r"https?://\S+", post)
    return urls_found

# Test — should find URLs only in the last two new posts
for post in posts_with_urls:
    urls = extract_urls(post)
    if urls:
        print(f"Found: {urls}")
````

```python
# Mentions
def extract_mentions_re(post):
    return [m.lower() for m in re.findall(r"@(\w+)", post)]

# Hashtags
def extract_hashtags_re(post):
    return [h.lower() for h in re.findall(r"#(\w+)", post)]

# URLs
def extract_urls(post):
    return re.findall(r"https?://\S+", post)
```

**Breaking down the URL pattern** `https?://\S+`:
- `http` — literal
- `s?` — the 's' is optional (matches both http and https)
- `://` — literal
- `\S+` — one or more non-whitespace characters (the rest of the URL)

**Why groups matter:** In `r"@(\w+)"`, the parentheses form a **capture group** — `findall()` returns only what's inside the group (the username), not the full match including `@`. Without the group, you'd get `['@alice', '@bob']` and have to strip the `@` yourself.
