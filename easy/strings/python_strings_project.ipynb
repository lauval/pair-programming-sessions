<details><summary>💡 Show regex patterns</summary>

```python
# Mentions
def extract_mentions_re(post):
    return [m.lower() for m in re.findall(r"@(\w+)", post)]

# Hashtags
def extract_hashtags_re(post):
    return [h.lower() for h in re.findall(r"#(\w+)", post)]

# URLs
def extract_urls(post):
    return re.findall(r"https?://\S+", post)
```

**Breaking down the URL pattern** `https?://\S+`:
- `http` — literal
- `s?` — the 's' is optional (matches both http and https)
- `://` — literal
- `\S+` — one or more non-whitespace characters (the rest of the URL)

**Why groups matter:** In `r"@(\w+)"`, the parentheses form a **capture group** — `findall()` returns only what's inside the group (the username), not the full match including `@`. Without the group, you'd get `['@alice', '@bob']` and have to strip the `@` yourself.
